## Question 1: Error Analysis Using Python

Write a Python program that accepts the true value and the approximate value of a quantity. The program should:

1. Compute the absolute error.
2. Compute the relative error.
3. Compute the percentage error.
4. Display the results with appropriate labels.
5. Determine the number of correct significant digits in the approximate value.

```python
import math
```

# Q1) Absolute Error

The absolute error is simply the magnitude of the gap between the true value and the approximate value:

$$E_a = |x_{true} - x_{approx}|$$

```python
def absolute_error(true_value, approx_value):
    return abs(true_value - approx_value)
```

# Q2) Relative Error

The relative error scales the absolute error by the size of the true value, so it's meaningful across different magnitudes:

$$E_r = \frac{|x_{true} - x_{approx}|}{|x_{true}|}, \quad x_{true} \neq 0$$

If the true value is 0, the relative error is undefined (treated here as infinite).

```python
def relative_error(true_value, abs_err):
    if true_value == 0:
        return float("inf")
    return abs_err / abs(true_value)
```

# Q3) Percentage Error

Just the relative error expressed as a percentage:

$$E_p = E_r \times 100\%$$

```python
def percentage_error(rel_err):
    return rel_err * 100
```

# Q5) Correct Significant Digits

A standard numerical-analysis criterion: the approximate value is said to be correct to $n$ significant digits if

$$E_r \leq 0.5 \times 10^{-n+1}$$

Solving for the largest such $n$ gives:

$$n = \left\lfloor 1 - \log_{10}(2 \, E_r) \right\rfloor$$

```python
def significant_digits(rel_err):
    if rel_err == 0:
        return float("inf")   # exact match
    if rel_err >= 1:
        return 0
    n = math.floor(1 - math.log10(2 * rel_err))
    return max(n, 0)
```

# Q4) Putting It All Together

Accept the true and approximate values, compute all four quantities, and display them with clear labels.

```python
def error_analysis(true_value, approx_value):
    abs_err = absolute_error(true_value, approx_value)
    rel_err = relative_error(true_value, abs_err)
    pct_err = percentage_error(rel_err)
    sig_digits = significant_digits(rel_err)
    return abs_err, rel_err, pct_err, sig_digits


def display_results(true_value, approx_value, abs_err, rel_err, pct_err, sig_digits):
    print("=" * 50)
    print("            ERROR ANALYSIS RESULTS")
    print("=" * 50)
    print(f"True Value                 : {true_value}")
    print(f"Approximate Value          : {approx_value}")
    print(f"Absolute Error              : {abs_err:.6f}")
    print(f"Relative Error              : {rel_err:.6f}")
    print(f"Percentage Error            : {pct_err:.4f} %")
    print(f"Correct Significant Digits  : {sig_digits}")
    print("=" * 50)
```

Now we call the program interactively, accepting the true and approximate values from the user.

```python
true_value = float(input("Enter the true value: "))
approx_value = float(input("Enter the approximate value: "))

abs_err, rel_err, pct_err, sig_digits = error_analysis(true_value, approx_value)
display_results(true_value, approx_value, abs_err, rel_err, pct_err, sig_digits)
```

# Demonstration

```python
true_value = math.pi
approx_value = 3.14159

abs_err, rel_err, pct_err, sig_digits = error_analysis(true_value, approx_value)
display_results(true_value, approx_value, abs_err, rel_err, pct_err, sig_digits)
```
